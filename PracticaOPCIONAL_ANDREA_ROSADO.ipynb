#### 2.- Dada la tupla _t_, convierte la tupla en un set. Calcula la suma de todos los elementos únicos y convierte el set de nuevo en tupla ordenada de mayor a menor.

```python
t = (4, 8, 15, 16, 23, 45, 6, 7, 9, 5, 111)
```

```python

# Convertir la tupla en un conjunto
conjunto = set(t)

# Sumar los elementos únicos
suma = sum(conjunto)

# Ordenar de mayor a menor y convertir en tupla
tupla_ordenada = tuple(sorted(conjunto, reverse=True))

# Mostrar resultados
print("Conjunto:", conjunto)
print("Suma:", suma)
print("Tupla ordenada:", tupla_ordenada)
```

#### 3.- Dada la cadena _texto_, crea un diccionario donde la clave sea cada letra y el valor sea la cantidad de veces que aparece. Convierte el diccionario en una lista de tuplas (letra, frecuencia) ordenada de mayor a menor frecuencia.

```python
texto = "banana"
```

```python

# Crear un diccionario para contar las letras
frecuencias = {}

for letra in texto:
    frecuencias[letra] = frecuencias.get(letra, 0) + 1

# Mostrar el diccionario
print("Diccionario:", frecuencias)

# Convertirlo en lista de tuplas ordenada por frecuencia
lista_ordenada = sorted(
    frecuencias.items(),
    key=lambda elemento: elemento[1],
    reverse=True
)

print("Lista ordenada:", lista_ordenada)
```

#### 4.- Dada la lista de tuplas _pares_. Filtra solo las tuplas donde el primer número es mayor que el segundo. Devuelve a una lista con True o False indicando si la suma de cada tupla es par.

```python
pares = [(1,5),(2,2),(10,3),(4,4)]
```

```python

# Filtrar las tuplas donde el primer número es mayor
filtradas = []

for a, b in pares:
    if a > b:
        filtradas.append((a, b))

# Comprobar si la suma de cada tupla filtrada es par
resultados = []

for a, b in filtradas:
    resultados.append((a + b) % 2 == 0)

print("Tuplas filtradas:", filtradas)
print("Resultados:", resultados)
```

#### 6.- Dado el string _frase_, obtén un set con todas las palabras únicas. Crea un diccionario que cuente cuántas veces aparece cada palabra. Devuelve el diccionario pero solo con las palabras que aparezcan más de 1 vez.

```python
frase = "En el camino, uno se da cuenta de que en un coche, uno siempre está en un compartimento, y, porque está acostumbrado, no se da cuenta de que, a través de esa ventana, todo lo que ve es simplemente más televisión"
```

```python

# Eliminar las comas y convertir todo a minúsculas
texto = frase.lower().replace(",", "")

# Separar el texto en palabras
palabras = texto.split()

# Obtener las palabras únicas
palabras_unicas = set(palabras)

# Crear un diccionario con las frecuencias
frecuencias = {}

for palabra in palabras:
    frecuencias[palabra] = frecuencias.get(palabra, 0) + 1

# Seleccionar solo las palabras repetidas
repetidas = {}

for palabra, cantidad in frecuencias.items():
    if cantidad > 1:
        repetidas[palabra] = cantidad

# Mostrar resultados
print("Palabras únicas:", palabras_unicas)
print("Palabras repetidas:", repetidas)
```

#### 7.- Dada la lista de tuplas _nums_, crea una lista con la diferencia entre el mayor y el menor de cada tupla. A la vez, genera una lista de True o False indicando si la suma de cada tupla es múltiplo de 3.

```python
nums = [(3,9),(10,2),(5,5),(8,1)]
```

```python

# Crear dos listas vacías
diferencias = []
multiplos_de_3 = []

# Recorrer las tuplas
for a, b in nums:

    # Calcular la diferencia entre el mayor y el menor
    diferencia = max(a, b) - min(a, b)
    diferencias.append(diferencia)

    # Comprobar si la suma es múltiplo de 3
    es_multiplo = (a + b) % 3 == 0
    multiplos_de_3.append(es_multiplo)

# Mostrar resultados
print("Diferencias:", diferencias)
print("Múltiplos de 3:", multiplos_de_3)
```

#### 9.- Dada la lista _valores_, convierte cada número a string, pero con las siguientes reglas:
- Si es par: "par:<num>"
- Si es múltiplo de 4: "mult3:<num>"
- Si cumple ambas condiciones, concatenar: "par-mult3:<num>"
#### Devuelve la lista final.

```python
valores = [10,21,33,42,55]
```

```python

# Crear una lista vacía
resultado = []

# Recorrer todos los números
for numero in valores:

    if numero % 2 == 0 and numero % 4 == 0:
        resultado.append("par-mult3:" + str(numero))

    elif numero % 2 == 0:
        resultado.append("par:" + str(numero))

    elif numero % 4 == 0:
        resultado.append("mult3:" + str(numero))

    else:
        resultado.append(str(numero))

# Mostrar la lista final
print(resultado)
```

#### 10.- Dado el dicionario _datos_, pon la primera letra de "nombre" en mayúsculas. Calcula la suma total de "puntos" y guarda el total en una nueva clave "total". Marca un booleano como "aprobado" si el total es mayor a 50. Finalmente, crea un set con todos los nombres de los usuarios aprobados.

```python
datos = [
    {"id":1,"nombre":"ana","puntos":(10,20,30)},
    {"id":2,"nombre":"luis","puntos":(5,15)},
    {"id":3,"nombre":"marta","puntos":(100,50,25)}
]
```

```python

# Crear un conjunto vacío para los aprobados
aprobados = set()

# Recorrer todos los usuarios
for usuario in datos:

    # Poner la primera letra en mayúscula
    usuario["nombre"] = usuario["nombre"].capitalize()

    # Calcular la suma de sus puntos
    usuario["total"] = sum(usuario["puntos"])

    # Comprobar si ha aprobado
    usuario["aprobado"] = usuario["total"] > 50

    # Añadir su nombre si ha aprobado
    if usuario["aprobado"]:
        aprobados.add(usuario["nombre"])

# Mostrar los resultados
print("Datos actualizados:")
for usuario in datos:
    print(usuario)

print("Usuarios aprobados:", aprobados)
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}
```

```python

# Diccionario original
estudiantes = {
    "s1": {"nombre": "ana", "notas": [10, 20, 30]},
    "s2": {"nombre": "luis", "notas": [15, 5]},
    "s3": {"nombre": "marta", "notas": [50, 100, 25]},
    "s4": {"nombre": "pepe", "notas": [40]}
}

# Crear un diccionario vacío
resultado = {}

# Recorrer todos los estudiantes
for estudiante in estudiantes.values():

    # Convertir el nombre a mayúsculas
    nombre = estudiante["nombre"].upper()

    # Obtener las notas
    notas = estudiante["notas"]

    # Calcular el promedio
    promedio = sum(notas) / len(notas)

    # Comprobar si ha aprobado
    aprobado = promedio >= 20

    # Obtener la nota máxima y mínima
    max_min = (max(notas), min(notas))

    # Guardar los resultados
    resultado[nombre] = {
        "promedio": promedio,
        "aprobado": aprobado,
        "max_min": max_min
    }

# Mostrar el nuevo diccionario
print(resultado)
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python

# Crear el archivo con las notas
with open("notas_alumnos.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Ana,10,20,30\n")
    archivo.write("Luis,5,15\n")
    archivo.write("Marta,100,50,25\n")

print("Archivo notas_alumnos.txt creado correctamente")
```

```python

# Lista donde guardaremos los estudiantes
resultados = []

# Abrir el archivo para leerlo
with open("notas_alumnos.txt", "r", encoding="utf-8") as archivo:

    for linea in archivo:

        # Separar los datos por comas
        datos = linea.strip().split(",")

        # Obtener el nombre
        nombre = datos[0].capitalize()

        # Convertir las notas a números
        notas = [int(nota) for nota in datos[1:]]

        # Calcular el promedio
        promedio = sum(notas) / len(notas)

        # Comprobar si está aprobado
        aprobado = promedio >= 20

        # Crear el diccionario
        estudiante = {
            "nombre": nombre,
            "promedio": promedio,
            "aprobado": aprobado
        }

        resultados.append(estudiante)

print(resultados)
```

```python

import json

# Guardar la lista en un archivo JSON
with open("resultado.json", "w", encoding="utf-8") as archivo:
    json.dump(resultados, archivo, indent=4, ensure_ascii=False)

print("Archivo resultado.json creado correctamente")
```

#### 15.- Con el archivo _fragmento-libro.txt_, ábrelo, cuenta la frecuencia de cada palabra (ignorando mayúsculas/minúsculas y signos). Devuelve un archivo llamado _frecuencias.json_ con las 5 palabras más comunes.

```python

from google.colab import files

archivos = files.upload()
```

```python

import os

print(os.path.exists("fragmento-libro.txt"))
```

```python

import re
from collections import Counter

# Abrir y leer el archivo
with open("fragmento-libro.txt", "r", encoding="utf-8") as archivo:
    texto = archivo.read()

# Convertir el texto a minúsculas
texto = texto.lower()

# Separar las palabras ignorando signos de puntuación
palabras = re.findall(r"\b\w+\b", texto)

# Contar las palabras
frecuencias = Counter(palabras)

# Obtener las 5 palabras más frecuentes
top_5 = frecuencias.most_common(5)

print(top_5)
```

```python

import json

# Convertir los resultados en un diccionario
resultado = dict(top_5)

# Guardar el archivo JSON
with open("frecuencias.json", "w", encoding="utf-8") as archivo:
    json.dump(resultado, archivo, indent=4, ensure_ascii=False)

print("Archivo frecuencias.json creado correctamente")
```

#### 16.- Crea tres archivos CSV (_ventas1.csv_, _ventas2.csv_, _ventas3.csv_ con la misma información:
- producto,cantidad
- A,10
- B,5
#### Combínalos en un único archivo que se llema _total_ventas.csv_. Suma las cantidades por producto (si el producto se repite).


```python
import csv

# Crear los tres archivos CSV
for i in range(1, 4):
    nombre_archivo = f"ventas{i}.csv"

    with open(nombre_archivo, "w", newline="", encoding="utf-8") as archivo:
        escritor = csv.writer(archivo)

        escritor.writerow(["producto", "cantidad"])
        escritor.writerow(["A", 10])
        escritor.writerow(["B", 5])

print("Los tres archivos CSV se han creado correctamente")
```

```python
# Diccionario para guardar las cantidades totales
totales = {}

# Leer los tres archivos
for i in range(1, 4):
    nombre_archivo = f"ventas{i}.csv"

    with open(nombre_archivo, "r", encoding="utf-8") as archivo:
        lector = csv.DictReader(archivo)

        for fila in lector:
            producto = fila["producto"]
            cantidad = int(fila["cantidad"])

            if producto in totales:
                totales[producto] += cantidad
            else:
                totales[producto] = cantidad

print("Cantidades totales:", totales)
```

```python
# Guardar las cantidades totales en un nuevo CSV
with open("total_ventas.csv", "w", newline="", encoding="utf-8") as archivo:
    escritor = csv.writer(archivo)

    escritor.writerow(["producto", "cantidad"])

    for producto, cantidad in totales.items():
        escritor.writerow([producto, cantidad])

print("Archivo total_ventas.csv creado correctamente")
```

```python
# Leer el archivo final para comprobarlo
with open("total_ventas.csv", "r", encoding="utf-8") as archivo:
    print(archivo.read())
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


```python
import json

# Crear la lista de alumnos
alumnos = [
    {"nombre": "ana", "notas": [10, 20, 30]},
    {"nombre": "luis", "notas": [5, 15]},
    {"nombre": "marta", "notas": [100, 50, 25]}
]

# Guardar los datos en un archivo JSON
with open("alumnos.json", "w", encoding="utf-8") as archivo:
    json.dump(alumnos, archivo, indent=4, ensure_ascii=False)

print("Archivo alumnos.json creado correctamente")
```

```python
# Abrir y leer el archivo JSON
with open("alumnos.json", "r", encoding="utf-8") as archivo:
    datos_alumnos = json.load(archivo)

print(datos_alumnos)
```

```python
# Crear un diccionario vacío
resultados = {}

# Recorrer todos los alumnos
for alumno in datos_alumnos:

    nombre = alumno["nombre"]
    notas = alumno["notas"]

    # Calcular el promedio
    promedio = sum(notas) / len(notas)

    # Calcular la nota máxima y mínima
    nota_maxima = max(notas)
    nota_minima = min(notas)

    # Comprobar si está aprobado
    aprobado = promedio >= 20

    # Guardar los resultados
    resultados[nombre] = {
        "promedio": promedio,
        "max": nota_maxima,
        "min": nota_minima,
        "aprobado": aprobado
    }

print(resultados)
```
